Loading image

```python
data_root_path = Path(os.path.expanduser("~")) / "Documents/Python Scripts/Infer-subc-2D"


in_data_path = Path("Z:/Cohen Lab/Maria Clara/2_Lab data/1_Multispectral data/2026/iAstrocytes Differentiation/iAstrocytes Differentiation - Deconvolved/tiff/iN day21")
out_data_path = Path("Z:/Cohen Lab/Maria Clara/2_Lab data/09_2026_Napari/iAstrocytes Napari/Mask cell")
im_type = ".tiff"

img_file_list = list_image_files(in_data_path,im_type)
```

BATCH CODE

```python
suffix="cell"
img_file_list = list_image_files(in_data_path, im_type)
for img_f in img_file_list:
    img_data, meta_dict = read_czi_image(img_f)
    agg1 = make_aggregate(img_in=img_data,
                     weight_ch0 = 0,
                     weight_ch1= 0,
                     weight_ch2= 10,
                     weight_ch3= 10,
                     weight_ch4= 0,
                     weight_ch5= 5,
                     weight_ch6= 20,
                     weight_ch7= 0,
                     weight_ch8= 0,
                     weight_ch9= 0,
                     rescale= True)
    sns = scale_and_smooth(img_in=agg1, 
                       median_size= 0, 
                       gauss_sigma= 10, 
                       slice_by_slice= True)
    #nlct = non_linear_cellmask_transform(sns)
    mot = masked_object_thresh(structure_img_smooth= sns, 
                           global_method='ave_tri_med', 
                           cutoff_size=2000, 
                           local_adjust=0.01)
    fafls = fill_and_filter_linear_size(img=mot, 
                                    hole_min=0, 
                                    hole_max=80, 
                                    min_size=5, 
                                    method= "3D", 
                                    connectivity= 1)
    hole_min_width = 0
    hole_max_width = 80
    small_object_width = 10
    fill_filter_method = "3D"
    cleaned_img2 = fill_and_filter_linear_size(fafls, 
                                           hole_min=hole_min_width, 
                                           hole_max=hole_max_width, 
                                           min_size= small_object_width,
                                           method=fill_filter_method)
    #viewer.add_image(cleaned_img2, scale=scale)
    export_inferred_organelle(cleaned_img2, suffix, meta_dict=meta_dict, out_data_path=out_data_path)
```
